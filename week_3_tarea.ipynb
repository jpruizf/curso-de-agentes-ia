```python
from langchain.agents import create_agent
from langchain.tools import tool

SYSTEM = ("Sos un analista que contesta sobre informes anuales. Citá el id del fragmento entre "
          "corchetes cuando uses buscar_en_informes. Si no alcanza, decilo. Contestá en pocas líneas.")

@tool
def buscar_dato(ticker: str, campo: str) -> str:
    '''TODO: escribí el docstring. Qué devuelve, qué campos existen, en qué unidad, para qué NO sirve.'''
    '''
    "Devuelve UN dato financiero del último ejercicio anual de una empresa, tomado de sus estados "
    "contables presentados ante la SEC y validado contra XBRL. Solo cubre estas 14 empresas: "
    "AAPL, MSFT, GOOG, AMZN, META, TSLA, JPM, JNJ, XOM, WMT, MELI, YPF, GLOB, IRS. "
    "Los importes están en millones de la moneda del balance (campo 'currency'). "
    "Para comparar dos empresas, llamala una vez por empresa. "
    "No sirve para preguntas sobre el texto del informe (estrategia, riesgos, explicaciones): "
    "para eso está buscar_en_informes.
    '''
    return lab_tools.buscar_dato(ticker, campo)

@tool
def buscar_en_informes(query: str, ticker: str | None = None) -> str:
    '''TODO: idem. La consulta va en inglés.'''
    '''
    "Busca fragmentos de texto en las secciones narrativas (descripción del negocio, factores de "
    "riesgo, análisis de la gerencia) de los informes anuales presentados ante la SEC por las mismas "
    "14 empresas. Devuelve hasta 5 fragmentos con su id (por ejemplo MELI-0214) y su sección. "
    "Sirve para preguntas sobre lo que la empresa DICE: estrategia, riesgos, competencia, explicaciones. "
    "No devuelve cifras validadas: para un número usá buscar_dato. "
    "Escribí la consulta en inglés, con los términos que usaría el informe."
    '''

    return buscar_en_informes_crudo(query, ticker)

TOOLS = [buscar_dato, buscar_en_informes]

def contestar_sin_memoria(pregunta: str) -> str:
    # TODO: armá el agente con create_agent y contestá la pregunta, sin pasarle nada de lo anterior

    # Creamos el agente stateless (sin memoria previa) usando tu modelo y las herramientas definidas
    
    agente= create_agent(
        model="ollama:llama3.2:3b",
        tools=TOOLS,
        system_prompt=SYSTEM
    )
    # Invoco al agente pasando únicamente el mensaje con la pregunta actual
    resultado= agente.invoke({"menssages": [{"role": "user", "content": pregunta}]})
    # Retornamos el contenido del último mensaje (la respuesta del agente)
    return resultado["messages"][-1].content
# Ejecución sobre los turnos de la primera conversación de prueba
c = convs[0]
for t in c["turnos"]:
    print(f"· {t['pregunta']}\n  → {contestar_sin_memoria(t['pregunta'])[:140]}\n")
```

```python
# ✅ CHECK 1: sin memoria, los turnos que no se entienden solos tienen que fallar
c = convs[0]
fallan = []
for t in c["turnos"]:
    if t["tipo"] == "directa":
        continue
    ok, _ = evals_mem.verificar(contestar_sin_memoria(t["pregunta"]), t["verificar"])
    fallan.append(not ok)
assert len(TOOLS) == 2, "faltan tools"
assert any(fallan), "raro: sin memoria debería fallar al menos un turno de seguimiento"
print(f"OK. Sin memoria fallan {sum(fallan)} de {len(fallan)} turnos de seguimiento.")
```
