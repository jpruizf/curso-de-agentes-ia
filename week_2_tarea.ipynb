<p style="margin: 0 0 12px 0;">Escribí la clase <code style="background: #e2e8f0; color: #0f172a; padding: 1px 6px; border-radius: 4px; font-size: 0.92em;">IndiceVectorial</code>: en el constructor, una matriz con un vector normalizado por fragmento; en <code style="background: #e2e8f0; color: #0f172a; padding: 1px 6px; border-radius: 4px; font-size: 0.92em;">buscar</code>, el vector de la pregunta, el producto con la matriz y los <code style="background: #e2e8f0; color: #0f172a; padding: 1px 6px; border-radius: 4px; font-size: 0.92em;">k</code> fragmentos de mayor coseno. Vectorizar los 1.605 fragmentos cuesta cerca de USD 0,01.</p></div></div>

```python
from ollama import Client
import numpy as np


# Inicializamos el cliente de Ollama apuntando a tu servidor local
client_emb= Client(host='127.0.0.1:11434')

class IndiceVectorial:
    def __init__(self, fragmentos,texto_de=lambda f: f["text"]):
        self.fragmentos = fragmentos

        # 1. Extraer los textos de cada fragmento
        textos = [texto_de(f) for f in self.fragmentos]
        
        # 2. Vectorizar todos los textos en lote usando Ollama
        vectores = []
        tamano_lote = 50

        print(f"Iniciando vectorizacion con ollama ({len(textos)} fragmentos)...")


        for i in range(0, len(textos), tamano_lote):
            lote= textos[i:i + tamano_lote]
            respuesta= client_emb.embed(model="nomic-embed-text", input=lote)
            vectores.extend(respuesta.embeddings)

        # 3. Convertir a array de NumPy
        matriz_raw= np.array(vectores)

        # 4. Asegurar exactamente 512 columnas
        if matriz_raw.shape[1] > 512:
            matriz_raw= matriz_raw[:, :512]

        # 5. Normalizar la matriz fila por fila (norma L2)
        normas= np.linalg.norm(matriz_raw, axis= 1, keepdims=True)
        self.matriz= matriz_raw / normas

    def buscar(self, pregunta, k=5, ticker= None):
        #Vectorizar la pregunta utilizando Ollama
        resp_pregunta= client_emb.embed(model="nomic-embed-text", input=pregunta)
        vector_pregunta= np.array(resp_pregunta.embeddings[0])


        if len(vector_pregunta) > 512:
            vector_pregunta= vector_pregunta[:512]

        vector_pregunta= vector_pregunta / np.linalg.norm(vector_pregunta)
        #Calcular similitud coseno (producto punto con la matriz normalizada)
        coseno= np.dot(self.matriz, vector_pregunta)

        if ticker is not None:
            for i, f in enumerate(self.fragmentos):
                if f.get("ticker") != ticker:
                    coseno[i] = -np.inf

        indices_mejores= np.argsort(coseno)[::-1][:k]

        resultados= []
        for i in indices_mejores:
            if coseno[i] != -np.inf:
                frag = self.fragmentos[i].copy()
                frag["score"]= float(coseno[i])
                resultados.append(frag)

        return resultados
#Instanciar el índice con el dataset completo
indice_v1= IndiceVectorial(fragmentos)

    # Verificaciones y pruebas
print("Filas en la matriz:", indice_v1.matriz.shape[0])
print("Total de fragmentos en la variable:", len(fragmentos))

for r in indice_v1.buscar("¿Qué dice YPF sobre Vaca Muerta?", k=3):
    print(f"[{r['id']}] {r['section']} {r['score']:.3f} | {r['text'][:100]!r}")

```

<p style="margin: 0 0 12px 0;">En el corpus del bloque, con 3.356 fragmentos, la versión binaria sola pierde 10 puntos y el rescate los recupera. En el tuyo, con la mitad de fragmentos, puede no perder nada, y el rescate hasta puede perder alguna: reordena 50 candidatos con el coseno real, así que un fragmento que los bits ponían tercero puede terminar séptimo. Con 20 preguntas eso es un acierto de diferencia, que ya sabemos que no se mide. Lo que no se discute es el tamaño.</p></div></div>

```python
M = indice_v2.matriz
# la misma consulta que usa tu buscador: traducida al inglés
vector_de = {p["pregunta"]: np.array(embeddings.embed_query(traducir(p["pregunta"])), dtype="float32") for p in preguntas}
bits = np.packbits(M > 0, axis=1)

def hit_rate(score_de):
    return sum(any(indice_v2.fragmentos[i]["id"] in p["relevantes"]
                   for i in np.argsort(-score_de(p["pregunta"]))[:5]) for p in preguntas) / len(preguntas)

def score_binario(pregunta, rescatar=0):
    q = vector_de[pregunta]
    distancia = np.unpackbits(bits ^ np.packbits(q > 0), axis=1).sum(axis=1)
    if not rescatar:
        return -distancia.astype("float32")
    # TODO 8: los `rescatar` mejores según los bits, reordenados con el coseno real
    candidatos= np.argsort(distancia)[:rescatar]

    scores= np.full(len(M), -np.inf, dtype="float32")

    scores[candidatos]= M[candidatos] @ q

    return scores

print(f"float32  {M.nbytes / 1e6:.1f} MB  hit-rate {hit_rate(lambda q: M @ vector_de[q]):.0%}")
print(f"binario  {bits.nbytes / 1e6:.1f} MB  hit-rate {hit_rate(lambda q: score_binario(q)):.0%}")
print(f"+rescate {bits.nbytes / 1e6:.1f} MB  hit-rate {hit_rate(lambda q: score_binario(q, 50)):.0%}")
```

```python
# ✅ CHECK 8 (opcional)
una = preguntas[0]["pregunta"]
puntuados = int((score_binario(una, 50) > -1e8).sum())
assert puntuados == 50, f"TODO 8: el rescate tiene que puntuar los 50 mejores según los bits, no {puntuados}"
mejores_por_bits = np.argsort(np.unpackbits(bits ^ np.packbits(vector_de[una] > 0), axis=1).sum(axis=1))[:50]
assert np.argsort(-score_binario(una, 50))[0] in mejores_por_bits, "TODO 8: el primero del rescate tiene que salir de esos 50"
print(f"OK. El índice pasa de {M.nbytes / 1e6:.1f} MB a {bits.nbytes / 1e6:.1f} MB.")
```
